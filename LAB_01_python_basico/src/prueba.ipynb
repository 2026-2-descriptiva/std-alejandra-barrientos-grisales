```python
from pathlib import Path

import gzip
import pandas as pd

data_path = next(
    parent / "data" / "data.csv.gz"
    for parent in (Path.cwd(), *Path.cwd().parents)
    if (parent / "data" / "data.csv.gz").exists()
)

with gzip.open(data_path, mode="rt", newline="") as data_file:
    df = pd.read_csv(
        data_file,
        header=None,
        sep="\t",
        names=["letter", "value", "date", "codes", "metrics"],
    )

df
```
